```python
import pandas as pd
imdb=pd.read_csv('../Dataset/imdb-top-1000.csv')
delevires=pd.read_csv('../Dataset/deliveries.csv')
imdb.head()
delevires.head()
### grouping by generes
return_value =imdb.groupby('Genre').mean(numeric_only=True)
print(imdb.groupby('Genre').groups)
### by default groupby returns a groupby object which is an iterator and we need to use aggregation function to get the result
### this will group the dataframe by genre and return the mean of each column for each genre
### while using aggreation we need to make sure that we specify numberic only columns
### after dividing into groups it tries to calculate the mean of each column
print(return_value)
### this will return the mean of each column for each genre
### top3 by total earnin 
total_earnings = imdb.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3)  
print(total_earnings)
### director by popularity

director_popularity = imdb.groupby('Director').mean(numeric_only=True)['No_of_Votes'].sort_values   
print(director_popularity) 
```

multiindex in pandas
What is an index?
- an index is a label , the whole idea of using pandas is becuase we have labels
- in numpy we only have positioned indexes
- multindex allows us to have multiple levels of index on Dataframesrows or columns
Can we have multplie indexes in series
- Yes a series can also have multindex it is series because it only has 1 d data
What special feature or advantage does multindex give us
- With multiindex we can access the data using two level of indexes

```python
### creating a multiindex dataframe
import pandas as pd

data = {
    "City": ["Toronto", "Toronto", "Thunder Bay", "Thunder Bay"],
    "Year": [2024, 2025, 2024, 2025],
    "Sales": [100, 150, 80, 120]
}

df = pd.DataFrame(data)

df = df.set_index(["City", "Year"])

print(df)
### accessing the multiindex dataframe using index we use loc
df.loc[("Toronto", 2024)]
### if we have multiple indexes then we dont need to specify all of them we can mention the index we want to use
df.loc["Thunder Bay"]
```
